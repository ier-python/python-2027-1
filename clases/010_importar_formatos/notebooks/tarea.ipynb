# Parte 3 — Algo nuevo: el formato Parquet

**Parquet** es un formato binario y comprimido para guardar tablas. Comparado con CSV:
- Conserva los tipos de datos: las fechas siguen siendo fechas y el texto sigue siendo texto, sin usar `index_col`, `parse_dates` ni `astype`.
- Ocupa mucho menos espacio en disco.
- Se lee y escribe mucho más rápido.

Para usarlo, pandas necesita una librería adicional llamada `pyarrow`. Si no la has instalado, instálala con `uv` desde la terminal, parado en tu espacio de trabajo `011_tarea_formatos`:

```
uv add pyarrow
```
