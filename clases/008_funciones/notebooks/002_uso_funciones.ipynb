```python
fecha = "2020-03-21"

def reformatea_fecha(fecha):
    """Toma una cadena con la forma YYYY-MM-DD y la regresa en DD/MM/YYYY
    """
    anio, mes, dia = fecha.split("-")
    nueva_fecha = f"{dia}/{mes}/{anio}"
    return nueva_fecha
```

```python
reformatea_fecha("1978-03-10")
```

```python
otra_fecha = "2026-09-17"
otra_fecha = reformatea_fecha(otra_fecha)
otra_fecha
```
