## Una función útil: reformatear fechas

`reformatea_fecha` junta lo visto en el curso: `split`, desempacado y
f-strings. Recibe `YYYY-MM-DD` y regresa `DD/MM/YYYY`.

```python
fecha = "2020-03-21"

def reformatea_fecha(fecha):
    """Toma una cadena con la forma YYYY-MM-DD y la regresa en DD/MM/YYYY
    """
    anio, mes, dia = fecha.split("-")
    nueva_fecha = f"{dia}/{mes}/{anio}"
    return nueva_fecha
```

```python
reformatea_fecha("1978-03-10")
```

```python
otra_fecha = "2026-09-17"
otra_fecha = reformatea_fecha(otra_fecha)
otra_fecha
```

## Cuidado con las variables globales

La primera versión de `hola_mundo` imprime `fecha`… que **no** es argumento:
la toma del espacio global de la libreta. Funciona de chiripa.

::: {.callout-caution}
Si una función necesita un dato, pídelo como argumento. Depender de variables
globales hace que la función se comporte distinto según qué celdas hayas
ejecutado antes.
:::

```python
def hola_mundo(nombre):
    print(f"Hola {nombre}")
    print(fecha)
```

```python
hola_mundo("Guadalupe")
```

La versión corregida recibe `fecha` como argumento: ahora el control es de
quien llama.

```python
def hola_mundo(nombre, fecha):
    print(f"Hola {nombre}")
    # fecha = "hoy"
    print(fecha)
```

```python
hola_mundo("Ricardo",fecha)
```

```python
hola_mundo("Ricardo","maniana")
```
