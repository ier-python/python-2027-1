## `def`: tu primera función

`def` abre la definición, los paréntesis llevan los **argumentos**, el bloque
indentado es el cuerpo y `return` entrega el resultado. Definirla no la
ejecuta: hay que *llamarla*.

```python
def suma(a,b):
    c = a + b
    return c
```

```python
suma(1,2)
```

```python
a = 1
b = 2
suma(a,b)
```

```python
def resta(a,b):
    c = a - b 
    return c
resta(1,2)
```

## Docstring

La cadena de triple comilla justo después del `def` documenta la función.

::: {.callout-tip}
Coloca el cursor sobre el nombre de una función y presiona `Shift+TAB` (o usa
`help(resta_ab)`): Jupyter te muestra el docstring. Documenta hoy, agradece
mañana.
:::

```python
def resta_ab(a,b):
    """
    regresa a - b"""
    return a - b 
resta_ab(1,2)
```

```python
resta_ab(1,-10)
```

## Argumentos por nombre

Al llamar con `a=10, b=10` el orden deja de importar y la llamada se
auto-documenta. Compara las tres celdas siguientes.

```python
resta_ab(a=10, b = 10)
```

```python
resta_ab(b=2, a = 11)
```

```python
resta_ab(a=2, b = 11)
```

```python
resta_ab(a=2, b = 11)
```

La última llamada **falla a propósito**: `"s" - 1` no existe. Python no
revisa los tipos al definir la función, sino hasta ejecutarla — eso es ser un
lenguaje dinámico.

```python
resta_ab("s",1)
```
