## Valores por defecto

`elevar_potencia(a, b=2)` eleva al cuadrado si no le dices otra cosa. Puedes
llamarla con uno o dos argumentos, por posición o por nombre.

```python
def elevar_potencia(a,b=2):
    """Toma a y la eleva a la potencia b, si b no se especifica, eleva al 
    cuadrado"""
    return a**b
```

```python
elevar_potencia(2)
```

```python
elevar_potencia(2,b= 0)
```

```python
elevar_potencia(b=3,a = 3)
```

```python
cuadrado = elevar_potencia(a = 3)
cuadrado
```

## Otra vez las globales

`suma_a_cuadrado` usa `cuadrado`, que vive afuera de la función: la misma
trampa de la libreta anterior. Funciona porque `cuadrado` ya existe, pero la
función miente sobre lo que necesita.

```python
def suma(a,b):
    """
    Realiza la suma de a + b
    """
    return a + b
```

```python
def suma_a_cuadrado(a):
    """
    Realiza la suma de a + cuadrado
    """
    return a + cuadrado
```

```python
suma_a_cuadrado(1)
```

## Defaults con cadenas

Los valores por defecto no tienen que ser números. Observa cómo al nombrar
solo `a` o solo `b` decides cuál default conservas.

```python
def saluda_personaA_personaB(a = "German",b="Jos'e Manuel"):
    """persona a saluda a persona b"""
    print(f"{a} saluda a {b}")
```

```python
saluda_personaA_personaB(b = "Roberto")
```

```python
saluda_personaA_personaB(a = "Roberto")
```

```python
saluda_personaA_personaB("hola",b="HOla")
```

```python
saluda_personaA_personaB("Guadalupe")
```

## La regla del orden

Los argumentos con default van **al final**. Las dos definiciones que fallan
a propósito (`SyntaxError: parameter without a default follows parameter with
a default`) lo demuestran: primero los obligatorios, luego los opcionales.

```python
def saluda_personaA_personaB(b="Jos'e Manuel", a):
    """persona a saluda a persona b"""
    print(f"{a} saluda a {b}")
```

```python
def suma_numeros(a,b,c,d=1,e=2,f=3):
    return a + b + c + d +e + f
```

```python
suma_numeros(1,2,3)
```

```python
def suma_numeros(a,b=1,c,d=1,e=2,f=3):
    return a + b + c + d +e + f
```
