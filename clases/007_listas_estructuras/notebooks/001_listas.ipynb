## `list()` sobre otras colecciones

`list()` convierte tuplas, conjuntos y cadenas en listas. Fíjate qué pasa con
cada una: el conjunto pierde el orden y la cadena se deshace en caracteres.

::: {.callout-caution}
Los métodos que modifican la lista *en el lugar* —como `reverse()`— regresan
`None`: por eso la celda no imprime nada y hay que ver la lista en la celda
siguiente.
:::

```python
nueva_lista = list( ("1",2,3,4,"sdaasd") )
nueva_lista = list( {"1",2,3,4,"sdaasd"} )
nueva_lista = list( "sdaasd" )
nueva_lista
```

```python
nueva_lista.reverse()
```

```python
nueva_lista
```

## Construir una lista pieza a pieza

La lista `alumnes` empieza vacía y crece con los métodos:

- `append(x)` agrega al final
- `insert(i, x)` inserta en la posición `i`
- `remove(x)` quita la primera aparición del *valor* `x`
- `pop(i)` quita el elemento en la *posición* `i` y te lo regresa

```python
alumnes = []
```

```python
alumnes.append("Guadalupe")
alumnes
```

```python
alumnes.append("Emiliano")
```

```python
alumnes.insert(0,"Manuel")
```

```python
alumnes
```

```python
alumnes.append("JorGE")
```

```python
alumnes
```

```python
alumnes.remove("JorGE")
```

```python
alumnes
```

```python
alumnes.append("Luis B")
```

```python
alumnes
```

```python
alumnes.pop(-1)
```

```python
alumnes
```

```python
alumnes.append("Luis")
```

Dos maneras de quitar elementos:

- `remove(valor)` busca el **valor** y quita su primera aparición
- `pop(posición)` quita por **posición** y además te regresa el elemento

## Desempacado (*unpacking*)

Si el lado izquierdo tiene tantas variables como elementos tiene la lista,
Python reparte los valores en un solo paso.

```python
fecha, hora, temperatura, direccion, velocidad = datos.split(",")
```

```python
fecha
```

La celda siguiente **falla a propósito**: el número de variables no coincide
con el número de pedazos. `count(",")` te dice cuántos campos esperar.

```python
fecha, hora = datos.split(",")
```

```python
datos.count(",")
```

## `removeprefix`, `removesuffix` y `title`

Útiles para limpiar nombres de archivo. Fíjate en la celda que *no* hace
nada: `removeprefix` solo quita el texto si está al **inicio**.

```python
archivo = "memo_categoria_vivienda"
```

```python
archivo.removeprefix("memo_")
```

```python
archivo.removeprefix("_vivienda")
```

```python
archivo.removesuffix("_vivienda")
```

```python
archivo.title()
```

## Formato numérico en f-strings

Dentro de las llaves, después de `:`, va la especificación de formato:
`.2f` (dos decimales), `.2e` (notación científica), `,.2f` (separador de
miles).

```python
fecha, hora, temperatura, direccion, velocidad = datos.split(",")
```

```python
print(f"Hoy es {fecha} y son las {hora} y la temperatura es {float(temperatura)}")
```

```python
print(f"La temperatura es {float(temperatura):.2f}")
```

```python
print(f"La temperatura es {float(temperatura):.2e}")
```

```python
print(f"La temperatura es {float(temperatura)*100:,.2f}")
```

## `zfill`: rellenar con ceros

`zfill(4)` rellena por la izquierda hasta completar 4 caracteres: así se
nombran archivos que ordenan bien (`0001`, `0002`, …). Para un entero,
primero conviértelo con `str()`.

```python
temperatura.zfill(4)
```

```python
entero  = 24
```

```python
str(entero).zfill(4)
```
