## De C a Python

En otros lenguajes, el `for` se escribe con contador, condición e incremento,
y el bloque va entre llaves:

```c
for (i = 0; i <= 10; i = i + 1) {
    /* dentro del for */
}
```

En Python no existe nada de eso: se recorre directamente una colección y el
bloque se define con indentación.

## `for` e indentación

En Python el bloque del `for` no lleva llaves: lo define la **indentación**.
Todo lo indentado se repite; la primera línea sin indentar ya está afuera.

````python
for i in range(2):
    print(i)
    print(i+1)
    print(i+1)
    print("fin de la estructura")
print("hola")
````

````python
for i in range(10):
    for j in range(10):
        for k in range(10):
            print(i,j,k)
````

## `while`

Repite mientras la condición sea verdadera. La responsabilidad de que la
condición algún día sea falsa es tuya: si olvidas el `i = i + 1`, el ciclo no
termina nunca. Las condiciones se pueden combinar con `and` / `or`.

````python
i  = 0
while i <= 10:
    print(i)
    i = i + 1
````

````python
i  = 0
j = 0
while (i <= 10) and (j >= 0):
    print(i)
    i = i + 1
````

## `if` / `else`

Ejecuta un bloque u otro según la condición. `==` compara; no lo confundas
con `=`, que asigna.

````python
nombre = 'Meo'
apellido= "del Valle"

if nombre == "Memo":
    print("Te llamas memo")
else:
    print(f"Eres alguien mas, eres {nombre}")
````

````python
colores= ["rojo","azul","morado",'amarillo',"rosa"]
for color in colores:
    if color == "azul":
        print("salio azul")
````

## Filtrar una lista

Patrón clásico: recorres `variables` y te quedas solo con las que contienen
`"Ti_"`, acumulándolas en una lista nueva con `append`.

````python
variables = ["Ti_sala","ach","Ti_cafe","temperatura"]
interes = []
for variable in variables:
    if "Ti_" in variable:
        interes.append(variable)
interes
````

## Comprensión de listas

La misma operación en una sola línea:
`[variable for variable in variables if "Ti_" in variable]`.
Es el modo *pythonico* de construir una lista filtrada.

````python
[variable for variable in variables if "Ti_" in variable]
````

````python
[variable  for variable in variables if "Ti_" in variable]
````
