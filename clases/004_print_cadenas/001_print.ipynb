# print en Python

En Python hay al menos tres maneras de imprimir:

1. Estilo C
2. str.format()
3. f-strings

## Estilo C

```python
print('Hola mundo cruel')
```

```python
nombre = 'Memo'
edad = 48
antiguedad = 16.5
```

```python
print('%s tiene %d años' % (nombre, edad))
print('{} tiene {} años'.format(nombre, edad))
print(f'{nombre} tiene {edad} años')
```

```python
print('%s lleva %.1f años en el IER' % (nombre, antiguedad))
print('{} lleva {:.1f} años en el IER'.format(nombre, antiguedad))
print(f'{nombre} lleva {antiguedad:.2f} años en el IER')
```

```python
print('%s, %d años, %.1f de antigüedad' % (nombre, edad, antiguedad))
print('{}, {} años, {} de antigüedad'.format(nombre, edad, antiguedad))
print(f'{nombre}, {edad} años, {antiguedad} de antigüedad')
```

```python
print('%s ingresó en %d' % (nombre, 2026 - antiguedad))
print('{} ingresó en {:.0f}'.format(nombre, 2026 - antiguedad))
print(f'{nombre} ingresó en {2026 - antiguedad:.1f}')
```

```python
print('%s trabajó el %.1f%% de su vida' % (nombre, 100*antiguedad/edad))
print('{} trabajó el {:.1f}% de su vida'.format(nombre, 100*antiguedad/edad))
print(f'{nombre} trabajó el {antiguedad/edad:.1%} de su vida')
```
