## `read_csv` con índice de tiempo

Las dos opciones clave: `index_col="date"` usa esa columna como índice y
`parse_dates=True` la convierte a fechas de verdad (`DatetimeIndex`).
`info()` resume columnas, tipos y memoria — es lo primero que se revisa tras
importar.

```python
f = "../data/001_raw/ClimaLab_2025-08-05_2026-08-05.csv"
esolmet = pd.read_csv(f,index_col="date",parse_dates=True)
esolmet.info()
```

## `resample`: promedios mensuales

Con el índice de tiempo, `resample("ME")` agrupa por mes (*Month End*) y
`mean()` promedia cada grupo. Un año de minutos se vuelve 12 renglones — la
primera recompensa de haber construido bien el índice.

```python
esolmet.tdb.resample("ME").mean()
```
