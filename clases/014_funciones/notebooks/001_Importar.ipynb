## La función

`import_esolmet` recibe tres cadenas con valores por omisión: el mes (`"01"`), el prefijo de la
ruta y el año (`"2026"`). Las pega en un patrón como
`../data/001_raw/ejercicio_02_fechas/ClimaLab_2026-04*`, lo pasa a `glob` y lee todos los
archivos que coinciden. Esta vez `concat` va con `axis=0`: cada archivo trae las diez columnas
de un día y lo que queremos es apilar días. `sort_index()` ordena por fecha, porque `glob` no
garantiza el orden. El `try`/`except` cubre el caso en que el patrón no encuentra nada:
`concat` de una lista vacía falla y, en vez del error, la función imprime qué lista recibió.

Los meses se escriben con dos dígitos (`"04"`, no `4`) porque así están en los nombres de
archivo y el patrón es una cadena. Las líneas comentadas (`print(f)`, `archivos`, `info()`) son
el rastro de cómo se depuró la función en clase.

```python
# mes = "04"
# cadena = "../data/001_raw/ejercicio_02_fechas/ClimaLab_2026-04-08.csv"

def import_esolmet(mes="01",cadena="../data/001_raw/ejercicio_02_fechas/ClimaLab_",anio="2026"):
    f = cadena + anio + "-"  + mes + "*"
    # print(f)
    archivos = glob.glob(f)
    # archivos
    try:
        esolmet = pd.concat([pd.read_csv(archivo,index_col=0,parse_dates=True) for archivo in archivos ],axis=0,sort=True)
        esolmet = esolmet.sort_index()
        # esolmet.info()
        return esolmet
    except:
        print(f"archivos es {archivos}")
```

## Un DataFrame por mes

Cinco llamadas, cinco DataFrames. Sólo cambia `mes`; la ruta y el año usan el valor por
omisión.

```python
abril = import_esolmet(mes="04")
mayo = import_esolmet(mes="05")
junio = import_esolmet(mes="06")
julio = import_esolmet(mes="07")
agosto = import_esolmet(mes="08")
```
